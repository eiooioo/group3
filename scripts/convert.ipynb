```python
import os
import sqlite3
import pandas as pd

# Step out of the scripts/ folder to find master_datasets/
excel_path = "../master_datasets/all_workers.xlsx"
db_path = "../master_datasets/workers_data.db"

# 1. Read all sheets from the Excel file
excel_file = pd.ExcelFile(excel_path)
conn = sqlite3.connect(db_path)

# 2. Iterate through each sheet and convert to a table in the database
for sheet_name in excel_file.sheet_names:
    df = pd.read_excel(excel_file, sheet_name=sheet_name)

    # Clean table name (replaces spaces with underscores)
    table_name = sheet_name.strip().lower().replace(" ", "_")

    # Export to SQLite database
    df.to_sql(table_name, conn, if_exists="replace", index=False)
    print(
        f"Sheet '{sheet_name}' successfully ingested into table '{table_name}' ({len(df)} rows)."
    )

conn.close()
print(f"\nDatabase successfully generated at: {os.path.abspath(db_path)}")
```
